```python
import pandas as pd
import geopandas as gpd

# 데이터 불러오기
df_2021 = pd.read_csv('final_crisis_index_by_dong_2021.csv')
gdf_2021 = gpd.read_file('final_crisis_index_by_dong_2021.gpkg')

# RII_grade가 데이터에 없으므로, RII_top30_flag로 대체해서 표를 만듭니다.
columns_to_show = ['자치구명', '행정동명', 'RII', 'RII_top30_flag', 'CCI', 'CCI_rank', 'risk_type']
top_10_rii = df_2021.nlargest(10, 'RII')[columns_to_show]

print("--- RII 상위 10개 행정동 ---")
display(top_10_rii)

# 상위 10개 중 복합위기 최고위험지역(triple_high_flag == 1) 포함 여부
triple_high_dongs = df_2021[df_2021['triple_high_flag'] == 1]
top_10_in_triple_high = top_10_rii[top_10_rii['행정동명'].isin(triple_high_dongs['행정동명'])]

# RII_grade가 데이터에 없으므로, RII_top30_flag로 대체해서 표를 만듭니다.
columns_to_show = ['자치구명', '행정동명', 'RII', 'RII_top30_flag', 'CCI', 'CCI_rank', 'risk_type']
top_10_rii = df_2021.nlargest(10, 'RII')[columns_to_show]

print("--- RII 상위 10개 행정동 ---")
display(top_10_rii)

# 상위 10개 중 복합위기 최고위험지역(triple_high_flag == 1) 포함 여부
triple_high_dongs = df_2021[df_2021['triple_high_flag'] == 1]
top_10_in_triple_high = top_10_rii[top_10_rii['행정동명'].isin(triple_high_dongs['행정동명'])]

print("\n--- 상위 10개 중 복합위기 최고위험지역 포함 행정동 ---")
display(top_10_in_triple_high)
```

```python
# RII_grade >= 4 대신, RII 상위 30%(RII_top30_flag == 1)이면서 최고위험지역이 아닌 곳 찾기
high_rii_low_cci = df_2021[(df_2021['RII_top30_flag'] == 1) & (df_2021['triple_high_flag'] == 0)]

print("--- RII는 상위권이지만 최고위험지역이 아닌 예시 ---")
display(high_rii_low_cci[['자치구명', '행정동명', 'RII', 'CCI', 'IVI', 'CDI']])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 깨짐 방지 (Windows 기준)
plt.rc('font', family='Malgun Gothic') 

# RII_grade별 행정동 수 막대그래프
plt.figure(figsize=(8, 5))
sns.countplot(data=df_2021, x='RII_grade', palette='Blues')
plt.title('RII 등급별 행정동 수')
plt.show()

# 자치구별 평균 RII
gu_mean_rii = df_2021.groupby('자치구명')['RII'].mean().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
gu_mean_rii.plot(kind='bar', color='skyblue')
plt.title('자치구별 평균 RII')
plt.xticks(rotation=45)
plt.show()
```
